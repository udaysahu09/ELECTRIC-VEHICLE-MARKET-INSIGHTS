```python
import pandas as pd
from pathlib import Path

# Load the CSV from the data folder
data_path = Path("electric_vehicles_spec_2025.csv")

df = pd.read_csv(data_path)

# Inspect the dataset
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())
```

```python
# 1. Check missing values
print("Missing values in each column:")
print(df.isnull().sum())

# 2. Check duplicate records
print("\nTotal duplicate rows:", df.duplicated().sum())

# 3. Check data types
print("\nColumn data types:")
print(df.dtypes)

# 4. View summary statistics
display(df.describe(include="all").T)
```

```python
# 1. Inspect duplicate model names
duplicate_models = df[df.duplicated(subset=["model"], keep=False)]

print("Records with repeated model names:")
display(duplicate_models[["brand", "model"]].sort_values("model"))

# 2. Inspect fully duplicated rows
print("Fully duplicated rows:", df.duplicated().sum())

# 3. Count missing values
missing = df.isnull().sum()
print("\nColumns with missing values:")
print(missing[missing > 0].sort_values(ascending=False))

# 4. Check suspicious numeric values
print("\nVehicle length below 2000 mm:")
display(df[df["length_mm"] < 2000][["brand", "model", "length_mm"]])

print("\nNumber of cells above 2000:")
display(df[df["number_of_cells"] > 2000][["brand", "model", "number_of_cells"]])
```

```python
# Create a separate working copy
clean_df = df.copy()

# Remove extra spaces from text values
text_cols = clean_df.select_dtypes(include="object").columns

for col in text_cols:
    clean_df[col] = clean_df[col].str.strip()

# Check for fully duplicated rows
print("Fully duplicated rows:", clean_df.duplicated().sum())

# Check missing values
print("\nMissing values:")
print(clean_df.isnull().sum()[clean_df.isnull().sum() > 0])

# Save the cleaned dataset
from pathlib import Path

output_path = Path("../data/ev_market_cleaned.csv")
clean_df.to_csv(output_path, index=False)

print("\nCleaned dataset saved at:", output_path.resolve())
print("Final dataset shape:", clean_df.shape)
```

```python
top_brands = (
    clean_df["brand"]
    .value_counts()
    .head(10)
)

print("Top 10 manufacturers by number of listed records:")
display(top_brands)
```

```python
top_range = clean_df.nlargest(10, "range_km")[
    ["brand", "model", "range_km", "battery_capacity_kWh"]
]

display(top_range)
```

```python
brand_analysis = (
    clean_df.groupby("brand")
    .agg(
        vehicle_count=("model", "nunique"),
        avg_range_km=("range_km", "mean"),
        avg_efficiency=("efficiency_wh_per_km", "mean"),
        avg_battery_kwh=("battery_capacity_kWh", "mean")
    )
    .sort_values("avg_range_km", ascending=False)
)

display(brand_analysis.head(10).round(2))
```
